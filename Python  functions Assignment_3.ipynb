```python
# Q12. Write a function calculate_grade(marks) that creates a local variable grade and  assigns:  
# A -> 75 and above,  
# B -> 60-74,  
# C -> 40-59,  
# F -> below 40.  
# Display the grade.
Marks = float(input('Enter marks:'))
def calculate_grade(marks=Marks):
    if marks >=75:
        print('Grade: A')
    else :
        if 60<=marks<=74:
            print('Grade: B')
        else :
            if 40<=marks<=59:
                print('Grade: C')
            else :
                print('Grade:F')
                
calculate_grade()
```

```python
# Q15. Create a global variable college_name. Write two functions: display_college() - 
#     displays the college name;  change_college() - changes the college name. Use the appropriate variable scope.
college_name = str(input('Enter current college:'))
next_college = str(input('Enter changed college:'))

def display_college():
    print('Current college:',college_name)
def change_college():
    global college_name 
    college_name = next_college 
    print('Next college:',college_name)
    
display_college()
change_college()
```

```python
# Q16. Create a global variable student_count = 0. Write a function add_student() that increases the student
#      count by 1 using the global keyword. Call the function multiple times and display the count.

student_count = 0
def add_student():
    global student_count 
    student_count = student_count + 1
    print(student_count)

add_student()
add_student()
add_student()
add_student()
add_student()
```

```python
# Q17. Create a global variable website_status = "Offline". Write a function update_status() 
#    thatchanges the value to "Online" using the global keyword. 

website_status = 'Offline'
def update_status():
    global website_status
    website_status = 'Online'
    print(website_status)

update_status()
```

```python
# Q18. Write a function display_subjects(*subjects) that accepts any number of subject names and displays each subject using a loop. 

def display_subjects(*subjects):
    for Subjects in subjects:
        print(Subjects)

    
display_subjects('Mathmatics','Physics','Biology','Chemistry')
    
```
