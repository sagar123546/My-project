```python
#1. Write a lambda function to check whether a person is eligible to vote. The person should be 18 or older.
check = lambda age : "Eligible" if age >=18 else "Not eligible"
print(check(float(input("Enter age: "))))
```

```python
#2. Write a lambda function to return the larger of three numbers.
Larger = lambda x,y,z : max(x,y,z)
x = float(input('num_1- '))
y = float(input('num_2- '))
z = float(input('num_2- '))
print('larger numer will be: ',Larger(x,y,z))
```

```python
#3. Write a lambda function to check whether a number is positive, negative, or zero.
Check = lambda x : 'Positive' if x > 0 else 'Negative' if x < 0 else 'Zero'
print('Number is ',Check(float(input('Enter a num '))))
```

```python
#4. Write a lambda function to calculate the area of a rectangle using length and width.
area = lambda l,b : l*b
l = float(input('length of rectangle: '))
b = float(input('width of rectangle: '))
print('Area of rectangle is ',area(l,b))
```

```python
#5. Write a lambda function to calculate simple interest using principal, rate, and time.
intrest = lambda p,r,t : p*r*t/100
p = float(input('principal: '))
r = float(input('annual rate: '))
t = float(input('time: '))
print('Final intrest will be ',intrest(p,r,t),'/-')
```

```python
#6. Write a lambda function to convert Celsius temperature into Fahrenheit.
Fahrenite = lambda c : (c*9/5)+32
print('Temperature in fahrenite: ',Fahrenite(float(input('Temperature in celsius: '))))
```

```python
#7. Write a lambda function to assign grades: A for marks 80 or above, B for 60 or above, C for 40 or above, otherwise Fail.
marks_grades = lambda m : 'A' if m >=80 else 'B' if m >=60 else 'C'
print(marks_grades(float(input('Enter marks: '))),'Grade')
```

```python
#8. Write a lambda function to calculate the final price after applying a discount percentage to an original price.
final_price = lambda price,dis : price - price*dis/100
price = float(input('Original_price: '))
dis = float(input('Discount rate: '))
print('Final price will be ',final_price(price,dis),'/-')
```

```python
#9. Write a lambda function to check whether a number is divisible by both 3 and 5.
find = lambda x : "'divisible'" if x % 15 ==0 else "'Non divisible'"
print('Given number is ',find(float(input('Enter a number: '))),'by both 3 and 5.')
```

```python
#10. Write a lambda function to calculate the total price of an item after adding 18% GST.
Final_price = lambda price,gst :  price + price*gst/100
price = float(input('Original price: ')) 
gst = float(input('gst rate: '))
print('Final price will be ',Final_price(price,gst),'/-')
```
